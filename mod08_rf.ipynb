### Feature engineering

```python
def feature_engineering(df=heart):
    df['Male'] = (df.Sex == 'Male').astype(int)
    df['HealthyDiet'] = (df.Diet == 'Healthy').astype(int)
    df['UnhealthyDiet'] = (df.Diet == 'Unhealthy').astype(int)
    df[['BP1','BP2']] = df['Blood Pressure'].str.split('/', n=1, expand=True).astype(int)

feature_engineering()
```

### Create a basic random forest model on the training set.

```python
rf_basic = RFC(max_features=len(indep_vars),
               criterion='log_loss',
               n_estimators=50,
               max_depth=10,
               min_samples_leaf=1,
               bootstrap=True,
               oob_score=True,
               random_state=seed)
rf_basic.fit(X_train, y_train)
```

### We'll use this function to quickly calculate the log loss of a model.

```python
def log_loss(y_true, y_prob, eps=1e-15):
    y_true = np.array(y_true)
    y_prob = np.array(y_prob)
    
    # avoid log(0)
    y_prob = np.clip(y_prob, eps, 1 - eps)
    
    ll = -np.mean(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))
    return ll
```

### Compute the log loss on the training set.

```python
y_hat_basic_train = rf_basic.predict_proba(X_train)[:,1]
print('train log loss:', log_loss(y_train, y_hat_basic_train))
```

### Calculate the out-of-bag error (bootstrapping) on training data to estimate the log loss on the test set.

```python
y_hat_basic_oob = rf_basic.oob_decision_function_[:,1]
print('oob log loss:  ', log_loss(y_train, y_hat_basic_oob))
```

### Compute the log loss on the test set.

```python
y_hat_basic_test = rf_basic.predict_proba(X_test)[:,1]
print('test log loss: ', log_loss(y_test, y_hat_basic_test))
```

### Build an improved random forest model on this same set of data. 

You will be scored according to how well it performs on an unseen holdout set.

```python
# Documentation: https://scikit-learn.org/dev/modules/generated/sklearn.ensemble.RandomForestRegressor.html

rf_better = RFC(max_features='sqrt',
               n_estimators=200,
               max_depth=10,
               min_samples_leaf=5,
               criterion='log_loss',
               oob_score=True,
               random_state=42)
rf_better.fit(X_train, y_train)
```

### (1) Explain what "value" means in the decision tree plots.

The "value" in the decision tree plots represent the number of records in each node that are included after the training splits are applied to the data. The two values presented like this: [class0, class1] are representing the class of the node, for example, whether it is no heart attack or heart attack. The Gini value represents the imurity or confidence in the node, higher values means more impurity and vice versa.

### (2) How is “log loss” different from the log likelihood we defined in class?

Log loss is the negative average log likelihood and with log loss the smaller the value the better as it measures failures of class assignement. The log likelihood is the average of the log probablilities which are assigned to the correct class. In short both represent the same thing except the sign is flipped, maximizing log ikelihood is the same as minmizing log loss.

We can tell that the basic model is overfit before attempting to build any others, as the model lacks any constraints. It doesn't give a max_depth or min_sample_leaf, it is simple and could be described exactly as we define overfitting, too much complexity. When the trees begin to grow since there is no constraints to guide them they will try and be as specific and accurate in representing every inch of the data. Another reason we can notice overfitting here is that there will be less training log loss here than testing log loss.
